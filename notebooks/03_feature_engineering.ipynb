# Feature Engineering

Transform the raw dataset into a modelling-ready form:
- Create binary target `at_risk`
- Drop G1/G2/G3 (avoid data leakage)
- Add engineered features
- Save to `data/student-mat-engineered.csv`

```python
import sys, os
sys.path.insert(0, os.path.abspath(".."))

from src.data import load_data, engineer_features, DATA_DIR
```

```python
df_raw = load_data()
print(f"Raw shape: {df_raw.shape}")
print(f"Columns: {list(df_raw.columns)}")
```

```python
df = engineer_features(df_raw)
print(f"Engineered shape: {df.shape}")
print(f"\nNew columns added: parent_edu, alcohol, studytime_failures")
print(f"Columns dropped: G1, G2, G3")
print(f"\nTarget distribution:")
print(df["at_risk"].value_counts())
```

```python
df[["parent_edu", "alcohol", "studytime_failures", "at_risk"]].describe()
```

```python
out_path = DATA_DIR / "student-mat-engineered.csv"
df.to_csv(out_path, index=False)
print(f"Saved engineered dataset to {out_path}")
```
